#### Conditions 1 & 3
- Subcritical flow

```python
%%render
mu = 0.9
B = 3 * m
h_roof = 0 * m # Roof level
h_sill = -3 * m # Sill level
h_in = 0 * m # Upper water level
h_cr = (2/3) * (h_in - h_sill) + h_sill
h_out = h_cr + 0.5 * m
Delta_h = h_in - h_out
U = sqrt(2*g*Delta_h)
d = minimum(h_out, h_roof) - h_sill
A = B * d
Q = mu * U * d * B
```
